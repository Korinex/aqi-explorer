# Task 2: Replace abnormal symbols with NaN

```python
import numpy as np
df = df.replace(['??', '##'], np.nan)
```

“Symbols like ?? and ## are not real values. replace() turns them into NaN so that pandas treats them as missing values and isnull() and fillna() work on them.”

# Task 3: Missing values

```python
print(df.isnull().sum())
```

```python
df['PM2.5'] = df['PM2.5'].fillna(df['PM2.5'].mean())
df['PM10'] = df['PM10'].fillna(df['PM10'].mean())
print(df.isnull().sum())
```

“PM2.5 had X missing values and PM10 had Y before; both have 0 after.”

# Task 4: Describe

```python
df.describe()
```

"describe() shows count, mean, std, min, quartiles and max for each numeric column. AQI has 24,850 observations. The column with the most missing values is Xylene (count = 9,478, so about 15,372 are missing). PM2.5 and PM10 now have full counts of 24,850 because we filled them with their means.”

# Task 5: NumPy analysis

```python
aqi = df['AQI'].to_numpy()
print(type(aqi), aqi.shape)
```

```python
print('Mean:', np.mean(aqi))
print('Max:', np.max(aqi))
print('Min:', np.min(aqi))
print('Median:', np.median(aqi))
print('Std dev:', np.std(aqi))
print('90th percentile:', np.percentile(aqi, 90))
```

```python
print('Days above 300:', len(aqi[aqi > 300]))
print(aqi[aqi > 300][:10])
```

“We converted the AQI column to a NumPy array using to_numpy(). The mean AQI is 166.46, the median is 118, the minimum is 13 and the maximum is 2049. There are 3,675 days with AQI above 300 (Very Poor or Severe), which is about 14.8% of the 24,850 days. The 90th percentile is 341, so 90% of days have AQI at or below 341.”

# Task 6: Worst day for each city

```python
worst = df.loc[df.groupby('City')['AQI'].idxmax()][['City', 'Date', 'AQI']]
worst
```

```python
worst.sort_values('AQI', ascending=False).head(3)
```

“groupby(‘City’)[‘AQI’].idxmax() gives the row label of the highest AQI in each city, and df.loc[...] fetches those full rows. The worst day in the whole dataset was in Ahmedabad on 2018-02-19 with AQI 2049. The next worst were Guwahati (956 on 2019-08-13) and Gurugram (891 on 2017-08-30).”
