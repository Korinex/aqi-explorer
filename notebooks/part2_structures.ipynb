```python
from google.colab import drive
drive.mount('/content/drive')
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/AQI_Project/data/city_day.csv')
df.head()
```

```python
df = df.dropna(subset=['AQI'])
print(df.shape)
```

```python
### LIST + TUPLES
# Slice first 2000 rows
small_df = df.head(2000).copy()

# Create list of tuples (city, date, AQI)
records = []
for i in range(len(small_df)):
    records.append((small_df.iloc[i]['City'], small_df.iloc[i]['Date'], small_df.iloc[i]['AQI']))

# Show first/last 5
print("First 5:")
print(records[:5])
print("\nLast 5:")
print(records[-5:])

# Max AQI in list
highest_aqi_tuple = max(records, key=lambda x: x[2])
print(f"\nMax AQI: {highest_aqi_tuple[2]} ({highest_aqi_tuple[0]}, {highest_aqi_tuple[1]})")
```

```python
### STRING HANDLING
# Clean input + split date
user_input = "  ahmedabad   "
city = user_input.strip().title()

date = '2019-05-01'
year, month, day = date.split('-')
print("Parsed:")
print(city, year, month, day)

# Count letter
target_letter = 'a'
letter_count = city.lower().count(target_letter)
print(f"\n'{target_letter}' count in '{city}': {letter_count}")

# Reverse city
reversed_city = city[::-1]
print(f"Reversed: {reversed_city}")
```

```python
### SETS OF POLLUTED CITIES
# Build sets
pm25_high = set(df[df['PM2.5'] > 60]['City'])
pm10_high = set(df[df['PM10'] > 100]['City'])

# Set operations
print('Both:', pm25_high & pm10_high)
print('\nEither:', pm25_high | pm10_high)
print('\nOnly PM2.5:', pm25_high - pm10_high)
print('\nOnly PM10:', pm10_high - pm25_high)

# Set lengths
print(f"\nPM2.5 count: {len(pm25_high)}")
print(f"PM10 count: {len(pm10_high)}")
```

```python
### NESTED DICTIONARY
# Build {city: {month: [AQI]}}
aqi_dict = {}
small = df.head(2000)
for i in range(len(small)):
    city_name = small.iloc[i]['City']
    month_str = small.iloc[i]['Date'][:7]
    aqi_val = small.iloc[i]['AQI']
    aqi_dict.setdefault(city_name, {}).setdefault(month_str, []).append(aqi_val)

# Print city/month values
test_city = 'Ahmedabad'
test_month = '2015-02'
print(f"{test_city} {test_month} AQI:", aqi_dict.get(test_city, {}).get(test_month, "No Data"))

# Add city
new_city = 'TestCity'
new_month = '2023-01'
aqi_dict[new_city] = {new_month: [45.0, 52.0, 60.0]}
print(f"\nAdded {new_city}: {aqi_dict[new_city]}")

# Update values
aqi_dict[new_city][new_month] = [40.0, 42.0, 48.0]
print(f"Updated {new_city}: {aqi_dict[new_city][new_month]}")

# Delete city
if 'TestCity' in aqi_dict:
    del aqi_dict['TestCity']
    print("\nDeleted TestCity")

# Sorted cities
print("\nSorted:")
for city_sorted in sorted(aqi_dict.keys()):
    print(f"- {city_sorted}")

# Highest average AQI city
sample_averages = {}
for city, months in aqi_dict.items():
    all_aqis = []
    for month, values in months.items():
        all_aqis.extend(values)
    if all_aqis:
        sample_averages[city] = sum(all_aqis) / len(all_aqis)

highest_avg_city = max(sample_averages, key=sample_averages.get)
print(f"\nMax avg AQI city: {highest_avg_city} ({sample_averages[highest_avg_city]:.2f})")
```

```python
### REPEATED VALUES
# Count repetitions
count_dict = {}

for val in df['AQI_Bucket'].dropna():
    count_dict[val] = count_dict.get(val, 0) + 1

print("Counts:")
for category, count in count_dict.items():
    print(f"{category}: {count}")
```
