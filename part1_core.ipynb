```python
# Category function
```

```python
def get_aqi_category(aqi):
    if aqi <= 50:
        return 'Good'
    elif aqi <= 100:
        return 'Satisfactory'
    elif aqi <= 200:
        return 'Moderate'
    elif aqi <= 300:
        return 'Poor'
    elif aqi <= 400:
        return 'Very Poor'
    else:
        return 'Severe'

```

```python
test_values = [30, 75, 150, 250, 350, 450]
for value in test_values:
    print(value, "->", get_aqi_category(value))
```

```python
# Advice function
```

```python

advice = {
    "Good": "Air is clean. Enjoy outdoor activities.",
    "Satisfactory": "Fine for most people. Sensitive people should be careful.",
    "Moderate": "People with asthma or heart problems should avoid heavy outdoor work.",
    "Poor": "Avoid long outdoor exercise. Wear a mask.",
    "Very Poor": "Stay indoors as much as possible. Wear a mask outside.",
    "Severe": "Health emergency. Avoid going outside."
}


def get_advice(category):
    if category in advice:
        return advice[category]
    else:
        return "No advice available"

```

```python
print(get_advice("Poor"))
print(get_advice("Banana"))
```

```python
# Class
```

```python
class AirReport:

    def __init__(self, city, date, aqi):
        self.city = city
        self.date = date
        self.aqi = aqi
        self.category = get_aqi_category(aqi)

    def get_report(self):
        text = "----- AIR QUALITY REPORT -----\n"
        text = text + "City     : " + str(self.city) + "\n"
        text = text + "Date     : " + str(self.date) + "\n"
        text = text + "AQI      : " + str(self.aqi) + "\n"
        text = text + "Category : " + self.category + "\n"
        text = text + "Advice   : " + get_advice(self.category) + "\n"
        text = text + "------------------------------"
        return text

```

```python

report1 = AirReport("Delhi", "2019-05-01", 320)
print(report1.get_report())
```

```python
# Save to a file
```

```python
def save_report(report):
    file_path = '/content/drive/MyDrive/AQI_Project/aqi_report.txt'
    f = open(file_path, 'a')
    f.write(report.get_report() + "\n\n")
    f.close()

```

```python
save_report(report1)

f = open('/content/drive/MyDrive/AQI_Project/aqi_report.txt', 'r')
print(f.read())
f.close()
```

```python
def main():
    while True:
        print("\n1. Check AQI for a city and date")
        print("2. Exit")
        choice = input("Enter your choice: ")

        if choice == "1":
            city = input("Enter city: ").strip().title()
            date = input("Enter date (YYYY-MM-DD): ").strip()


            row = df[(df['City'] == city) & (df['Date'] == date)]

            if row.empty:
                print("No data found for this city and date.")
            else:
                aqi_value = row.iloc[0]['AQI']
                report = AirReport(city, date, aqi_value)
                print(report.get_report())
                save_report(report)
                print("Report saved in file.")

        elif choice == "2":
            print("Thank you!")
            break

        else:
            print("Wrong choice, try again.")

main()
```
